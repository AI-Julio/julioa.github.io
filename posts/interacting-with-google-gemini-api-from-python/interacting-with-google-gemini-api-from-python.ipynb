3. Interact with API by initializing genai.Client() below is the code you need to use.

```python
import os
from google.colab import userdata

# 1. Get the key value from Colab Secrets
api_key = userdata.get('GEMINI_API_KEY')

# 2. Assign the value to the environment variable
if api_key:
    os.environ['GEMINI_API_KEY'] = api_key
else:
    raise ValueError("GEMINI_API_KEY not found in Colab Secrets.")

# Now, the environment variable is set, and the client will initialize successfully.
from google import genai
client = genai.Client() # This should now work!
```

4. we are going to play with 3 different google genai prompts
A. Generate text prompt
B. Generate Images and Text responses
C. Generate Streaming responses.

1. Generate Text Prompts: below is the Code

```python
prompt = "Explain what the Book of Genesis is all about in a couple sentences"

response = client.models.generate_content(
    model='gemini-2.5-flash',
    contents=prompt,
)

print("--- Generated Text ---")
print(response.text)
```

2. Generate both images and text responses: Below is the Code

```python
from PIL import Image

# --- 1. Define the path to your actual image file ---
image_path = 'thearkofnoah.jpg' # <-- REPLACE with your file path!

# --- 2. Load the image directly from the file path ---
try:
    img_for_api = Image.open(image_path)
except FileNotFoundError:
    print(f"Error: Image file not found at {image_path}. Please check the path.")
    # Exit or handle the error appropriately
    # raise
    # return

# --- Define the prompt parts ---
contents = [
    "can you explain what is in the image and what you see?",
    # The loaded PIL Image object is passed directly
    img_for_api
]

# --- Generate content ---
# Assuming 'client' and 'model' are already defined
response = client.models.generate_content(
    model='gemini-2.5-flash',
    contents=contents,
)

print("\n--- Multimodal Response ---")
print(response.text)
```

3. Generate Streaming responses: Below is the Code

```python
prompt_long = "i want you to give me a one sheet fill in the blank sheet of questions for the book of Genesis 1"

print("\n--- Streaming Response ---")
response_stream = client.models.generate_content_stream(
    model='gemini-2.5-flash',
    contents=prompt_long,
)

for chunk in response_stream:
    # print the text chunk without a newline until the end
    print(chunk.text, end="", flush=True)

print("\n--- End of Stream ---")
```
